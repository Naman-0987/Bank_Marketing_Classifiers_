```python
X = bank.drop(columns=['y'])
y = bank['y']
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

Y_scaled = le.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

X_train,X_test,y_train,y_test = train_test_split(X,Y_scaled,test_size=.2,random_state=42)
```

```python
y_train
```

```python
from sklearn.preprocessing import OneHotEncoder

from sklearn.pipeline import make_pipeline

from sklearn.compose import ColumnTransformer

from sklearn.preprocessing import OrdinalEncoder 

from sklearn.preprocessing import StandardScaler
```

```python
education_order = [
    'illiterate', 
    'basic.4y', 
    'basic.6y', 
    'basic.9y', 
    'high.school', 
    'professional.course', 
    'university.degree'
]


transform = ColumnTransformer(transformers=[
    ('ore',OrdinalEncoder(categories=[education_order]),['education']),
    ('ohe',OneHotEncoder(drop='first', handle_unknown='ignore'),['job','marital','housing','loan','contact','month','day_of_week','poutcome']),
    ('sts',StandardScaler(),['age','duration','campaign','pdays','previous','emp.var.rate','cons.price.idx','cons.conf.idx','euribor3m','nr.employed','previously_contacted'])
    ],remainder='passthrough'
)

# Fit and transform the training and testing sets
X_train_transformed = transform.fit_transform(X_train)
X_test_transformed = transform.transform(X_test)
```

```python
from sklearn.model_selection import RandomizedSearchCV

clf = DecisionTreeClassifier(random_state=42)

# 1. Set up the parameter grid
param_grid = {
    "criterion":["gini", "entropy","log_loss"],
    "max_depth": [5,6,7,8,9,10],
    "min_samples_split": [5,6,7,8,9,10],
    "min_samples_leaf": [3,4,5],
    "splitter": ["best", "random"]
}

# 2. Initialize RandomizedSearchCV
random_search = RandomizedSearchCV(estimator=clf, param_distributions=param_grid, cv=5,scoring='accuracy',n_jobs=-1,random_state=42)

# 3. Fit to the training data
random_search.fit(X_train_transformed, y_train)

# View the best parameters found
print("Best Parameters:", random_search.best_params_)
print("Best Accuracy Score:", random_search.best_score_)
```

```python
model_pipeline = make_pipeline(
    transform,
    DecisionTreeClassifier(random_state=42,splitter='best',min_samples_split=9,min_samples_leaf=4,max_depth=7,criterion='entropy',class_weight='balanced')
)

# 3. Fit and score directly on raw X_train / X_test
model_pipeline.fit(X_train, y_train)
accuracy = model_pipeline.score(X_test, y_test)
print(f"Base Model Accuracy: {accuracy:.4f}")
```

```python
from sklearn.metrics import classification_report

y_pred = model_pipeline.predict(X_test)

# Print detailed classification metrics 
print(classification_report(y_test, y_pred, target_names=['no', 'yes']))
```

```python
from sklearn.metrics import roc_auc_score

# Get predicted probabilities for the positive class (class 1)
y_probs = model_pipeline.predict_proba(X_test)[:, 1]

# Calculate ROC-AUC Score
roc_auc = roc_auc_score(y_test, y_probs)
print(f"ROC-AUC Score: {roc_auc:.4f}")
```

```python
from sklearn.ensemble import VotingClassifier

# Define base models using your optimal tuned parameters
dt_tuned = DecisionTreeClassifier(
    random_state=42,
    splitter="best",
    min_samples_split=9,
    min_samples_leaf=4,
    max_depth=7,
    criterion="entropy",
    class_weight="balanced",
)

lr_tuned = LogisticRegression(
    random_state=42,
    solver="liblinear",
    C=10,
    
    class_weight=None,
)

svc_tuned = SVC(
    kernel="rbf",
    gamma="scale",
    C=0.1,
    class_weight=None,
    probability=True,
    random_state=42,
)

# Build Soft Voting Ensemble Pipeline
# Soft voting averages predicted probabilities across models
voting_clf = VotingClassifier(
    estimators=[("dt", dt_tuned), ("lr", lr_tuned), ("svc", svc_tuned)],
    voting="soft",
    n_jobs=4,
)

ensemble_pipeline = make_pipeline(transform, voting_clf)

# Fit on raw training data
ensemble_pipeline.fit(X_train, y_train)

# Predict and evaluate performance
y_pred_ensemble = ensemble_pipeline.predict(X_test)
y_probs_ensemble = ensemble_pipeline.predict_proba(X_test)[:, 1]

print("--- Ensemble Classification Report ---")
print(
    classification_report(
        y_test, y_pred_ensemble, target_names=["no", "yes"]
    )
)
print(f"ROC-AUC Score: {roc_auc_score(y_test, y_probs_ensemble):.4f}")
```
